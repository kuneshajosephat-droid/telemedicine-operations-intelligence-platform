```python
import pandas as pd

df = pd.read_csv("telehealth_data.csv")

# 👇 ADD IT HERE
df.columns = df.columns.str.lower()
df["telehealth_rate"] = df["telehealth_rate"] * 100

df.head()
```

```python
import matplotlib.pyplot as plt

# Fix column names (important)
df.columns = df.columns.str.lower()

# --- 1. Trend over time ---
trend = df.groupby("year")["telehealth_rate"].mean()

plt.figure()
trend.plot(marker='o')
plt.title("Telehealth Adoption Trend (USA)")
plt.xlabel("Year")
plt.ylabel("Average Telehealth Rate")
plt.grid()
plt.show()
```

```python
# --- 2. Top 10 regions ---
top_regions = (
    df.groupby("region")["telehealth_rate"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure()
top_regions.plot(kind='bar')
plt.title("Top 10 States by Telehealth Adoption")
plt.xticks(rotation=45)
plt.show()
```

```python
# --- 3. Lowest 10 regions ---
low_regions = (
    df.groupby("region")["telehealth_rate"]
    .mean()
    .sort_values()
    .head(10)
)

plt.figure()
low_regions.plot(kind='bar')
plt.title("Lowest 10 States by Telehealth Adoption")
plt.xticks(rotation=45)
plt.show()
```

```python
import fitz
from PIL import Image

doc = fitz.open("telemedicine_dashboard.pdf")

images = []

for i, page in enumerate(doc):
    pix = page.get_pixmap(matrix=fitz.Matrix(3, 3))
    img_path = f"page_{i}.png"
    pix.save(img_path)
    images.append(Image.open(img_path))

width = max(img.width for img in images)
height = sum(img.height for img in images)

combined = Image.new("RGB", (width, height), "white")

y = 0
for img in images:
    combined.paste(img, (0, y))
    y += img.height

combined.save("telemedicine_dashboard.png")

print("✅ Done — image created")
```

```python
import pandas as pd
import pickle
import os

from sklearn.ensemble import RandomForestRegressor
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline

# Load data
df = pd.read_csv("telehealth_data.csv")
df.columns = df.columns.str.lower().str.strip()

# Keep Overall records only
df = df[df["quarter"].astype(str).str.lower().str.strip() == "overall"].copy()

# Convert numeric columns
for col in ["year", "total_enrollment", "telehealth_users", "telehealth_rate"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Drop missing values
df = df.dropna(subset=[
    "year", "region", "total_enrollment",
    "telehealth_users", "telehealth_rate"
])

# Features and target
X = df[["year", "region", "total_enrollment", "telehealth_users"]]
y = df["telehealth_rate"]

# Model pipeline: encode region + train Random Forest
preprocessor = ColumnTransformer(
    transformers=[
        ("region_encoder", OneHotEncoder(handle_unknown="ignore"), ["region"])
    ],
    remainder="passthrough"
)

model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=200,
        random_state=42,
        max_depth=5
    ))
])

# Train
model.fit(X, y)

# Save
os.makedirs("models", exist_ok=True)

with open("models/telehealth_model.pkl", "wb") as f:
    pickle.dump(model, f)

print("✅ Model saved successfully")
print(os.listdir("models"))
```
